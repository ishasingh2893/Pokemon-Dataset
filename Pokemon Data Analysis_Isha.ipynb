```python
# Adding a column

# df['Total'] = df['HP'] + df['Attack'] + df['Defense'] + df['Sp. Atk'] + df['Sp. Def'] + df['Speed']

# Dropping a column

# df.drop(columns=['Total'])

df['Total'] = df.iloc[:, 4:10].sum(axis=1) 

# axis=1 adds entries row-wise, axis=1 adds entries col-wise
# notice that the first parameter in the iloc function is indicating to include all rows 
# the second parameter indicates to only include the columns 4 to 9; 10th is excluded

df.head(5)
```

```python
# Moving the 'total' column to the mid of the dataset
cols = list(df.columns)
df = df[cols[0:4] + [cols[-1]] + cols[4:12]]
df.head(5)
```

```python
# Save this updated CSV file

# df.to_excel('modified.xlsx', index=False) sacin
# df.to_csv('modified.txt', index=False, sep='\t')

df.to_csv('modified.csv', index=False) # removes the index col
```

Filtering of Data

```python
# df.loc[(df['Type 1'] == 'Grass') & (df['Type 2'] == 'Poison')]

# new_df = df.loc[(df['Type 1'] == 'Grass') | (df['Type 2'] == 'Poison')] # this is for 'or'

new_df = df.loc[(df['Type 1'] == 'Grass') & (df['Type 2'] == 'Poison') & (df['HP'] > 70)]
# this is a new df and wont change the OG df
# can save it in a new file
# new_df.to_csv('filtered.csv')
```

```python
# if we want to remove the names of pokemon containing the word 'mega'

# df.loc[df['Name'].str.contains('Mega')] # displays all the rows containing the word mega

df.loc[~df['Name'].str.contains('Mega')] # filters out the mega rows
```

Aggregate statistics (GroupBy)

```python
df = pd.read_csv('modified.csv')
```

```python
df.groupby('Type 1').mean(numeric_only=True).sort_values('Defense', ascending=False)
# We can change the name of the column in above code and sort the data from highest to lowest with ascending = False
```

```python
# df.groupby('Type 1').sum()

df.groupby('Type 1').count()
```

```python
# df.groupby('Type 1').size()

df.groupby('Type 1').size().sort_values(ascending=False)
```
